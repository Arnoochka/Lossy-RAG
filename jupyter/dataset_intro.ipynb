```python
import pandas as pd

splits = {'train': 'plain_text/train-00000-of-00001.parquet', 'valid': 'plain_text/validation-00000-of-00001.parquet'}
df_train = pd.read_parquet("hf://datasets/rajpurkar/squad/" + splits["train"])
df_valid = pd.read_parquet("hf://datasets/rajpurkar/squad/" + splits["valid"])
df = pd.concat([df_train, df_valid], ignore_index=True)
```

```python
df.head(5)
```

```python
unique = {'column': [], 'percentage': []}
for column in df.columns:
    try:
        unique['percentage'].append(len(df[column].unique()) / len(df))
        unique['column'].append(column)
    except TypeError:
        print(f"NOT: {column}")
    
pd.DataFrame(unique)
```
